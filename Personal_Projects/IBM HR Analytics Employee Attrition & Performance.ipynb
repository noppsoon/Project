#### <b> Group Comparison Statistics </b>

```python
def numeric_group_comparison(df, numeric_cols, target="Attrition"):

    summary = df.groupby(target)[numeric_cols].agg(["mean", "median"])

    comparison = pd.DataFrame(index=numeric_cols)

    for col in numeric_cols:

        no_mean = summary.loc["No", (col, "mean")]
        yes_mean = summary.loc["Yes", (col, "mean")]

        no_median = summary.loc["No", (col, "median")]
        yes_median = summary.loc["Yes", (col, "median")]

        comparison.loc[col, "No_mean"] = no_mean
        comparison.loc[col, "Yes_mean"] = yes_mean
        comparison.loc[col, "Mean_difference"] = yes_mean - no_mean
        comparison.loc[col, "Abs_mean_difference"] = abs(yes_mean - no_mean)

        comparison.loc[col, "No_median"] = no_median
        comparison.loc[col, "Yes_median"] = yes_median
        comparison.loc[col, "Median_difference"] = yes_median - no_median
        comparison.loc[col, "Abs_median_difference"] = abs(yes_median - no_median)

        # -------- Added percentage columns --------
        comparison.loc[col, "Mean_pct_difference"] = (
            (yes_mean - no_mean) / no_mean * 100
            if no_mean != 0 else None
        )

        comparison.loc[col, "Median_pct_difference"] = (
            (yes_median - no_median) / no_median * 100
            if no_median != 0 else None
        )

        comparison.loc[col, "Abs_mean_pct_difference"] = abs(
            comparison.loc[col, "Mean_pct_difference"]
        )

        comparison.loc[col, "Abs_median_pct_difference"] = abs(
            comparison.loc[col, "Median_pct_difference"]
        )

    return comparison.sort_values(
        "Abs_mean_pct_difference",
        ascending=False
    )
```

#### <b> Mann-Whitney U Test </b>

Is there a significant difference in this numeric feature between employees who left and those who stayed? 

<b> Hypotheses </b> 
<br> <b> Null hypothesis (H₀): </b> There is no significant difference in this numeric feature between employees who left and those who stayed.
<br> <b> Alternative hypothesis (H₁): </b> There is a significant difference in this numeric feature between employees who left and those who stayed.
<br>
<br> <b> Interpretation </b> :
<br> p < 0.05 → Reject H₀: the feature is significantly associated with attrition.
<br> p ≥ 0.05 → Fail to reject H₀: there is insufficient evidence of an association.

```python
from scipy.stats import mannwhitneyu

yes = df_more_1_val[df_more_1_val['Attrition'] == 'Yes']
no = df_more_1_val[df_more_1_val['Attrition'] == 'No']

results = []

for col in num_cols_more_1_val:
    stat, p = mannwhitneyu(yes[col], no[col])
    results.append({
        'Feature': col,
        'P-Value': round(p, 4),
        'Significant': 'Yes' if p < 0.05 else 'No'
    })

result_df = pd.DataFrame(results).sort_values('P-Value')
print(result_df)

# Get significant features separately
num_sig = result_df[result_df['Significant'] == 'Yes']['Feature'].tolist()

print("\nSignificant features:")
for col in num_sig:
    print(col)
```

<b> Statistical Confirmation (Mann-Whitney U & Correlation) </b>

The Mann–Whitney U test was conducted to determine whether numeric-feature distributions differed between employees who left and those who stayed. MonthlyIncome, Age, TotalWorkingYears, YearsAtCompany, YearsInCurrentRole, YearsWithCurrManager, JobLevel, and DistanceFromHome showed statistically significant differences (p < 0.05). In contrast, rate-based variables, PerformanceRating, PercentSalaryHike, and HourlyRate showed no significant differences. Point-biserial correlation further indicated that tenure-, income-, and age-related features were negatively associated with attrition, while DistanceFromHome had a weak positive association.

*** <b>Note:</b> Due to the large sample size, the Shapiro–Wilk test was highly sensitive to small departures from normality. Histograms and boxplots were therefore used to assess distribution shape, and the Mann–Whitney U test was used for group comparisons.

#### Numeric features EDA summary

Numeric feature analysis identified MonthlyIncome, TotalWorkingYears, YearsAtCompany, YearsInCurrentRole, YearsWithCurrManager, Age, and DistanceFromHome as the features most clearly related to employee attrition. These findings were supported by the boxplots, descriptive statistics, point-biserial correlations, and Mann–Whitney U tests. In contrast, HourlyRate, MonthlyRate, PerformanceRating, EmployeeNumber, and PercentSalaryHike showed little evidence of a meaningful relationship with attrition.

#### <b> Categorical Attrition Rate </b>

```python
import math

def plot_categorical_attrition_grid(df, cat_cols, target="Attrition", n_cols=3):
    
    # Calculate required subplot rows
    n_rows = math.ceil(len(cat_cols) / n_cols)
    
    # Create subplot grid
    _, axes = plt.subplots(
        n_rows, 
        n_cols, 
        figsize=(n_cols * 7, n_rows * 5))
    axes = axes.flatten()

    for i, col in enumerate(cat_cols):
        
        # Calculate attrition rate (%) for each category
        attrition_rate = (
            pd.crosstab(df[col], df[target], normalize="index")["Yes"]
            .mul(100)
            .sort_values(ascending=True)
        )

        # Highlight category with the highest attrition rate
        colors = [
            "orange" if value == attrition_rate.max() else "steelblue"
            for value in attrition_rate
        ]

        # Create horizontal bar chart
        axes[i].barh(
            attrition_rate.index.astype(str), 
            attrition_rate.values, 
            color=colors)
        
         # Set subplot labels
        axes[i].set_title(f"Attrition Rate by {col}")
        axes[i].set_xlabel("Attrition Rate (%)")
        axes[i].set_ylabel(col)

        for bar in axes[i].patches:
            width = bar.get_width()
            axes[i].text(width + 0.3, bar.get_y() + bar.get_height() / 2,
                         f"{width:.2f}%", va="center", fontsize=9)

    # Hide unused subplots
    for j in range(i + 1, len(axes)):
        axes[j].set_visible(False)

    plt.suptitle("Attrition Rate by Categorical Features", fontsize=16, y=1.02)
    plt.tight_layout()
    plt.show()
```

```python
# Usage
plot_categorical_attrition_grid(df_more_1_val, object_cols_more_1_val)
```

#### <b> Chi-Square Test of Independence </b>

Is there a significant association between this categorical feature and employee attrition?

<b> Hypotheses </b>
<br> <b> Null hypothesis (H₀): </b> There is no significant association between this categorical feature and employee attrition.
<br> <b> Alternative hypothesis (H₁): </b> TThere is a significant association between this categorical feature and employee attrition.
<br>
<br> <b> Interpretation </b> :
<br> p < 0.05 → reject H₀ → the feature is significantly associated with attrition.
<br>p ≥ 0.05 → Fail to reject H₀: there is insufficient evidence of an association.

```python
from scipy.stats import chi2_contingency

chi_results = []

for col in object_cols_more_1_val:
    if col != "Attrition":
        table = pd.crosstab(df_more_1_val[col], df_more_1_val["Attrition"])
        chi2, p, dof, expected = chi2_contingency(table)
        chi_results.append({
            'Feature': col,
            'P-Value': round(p, 4),
            'Significant': 'Yes' if p < 0.05 else 'No'
        })

chi_result_df = pd.DataFrame(chi_results).sort_values('P-Value')
print(chi_result_df)
```

The Chi-square test was conducted to examine whether categorical features were associated with employee attrition. BusinessTravel, JobRole, MaritalStatus, OverTime, Department, and EducationField all showed statistically significant association with attrition (p < 0.05), indicating that the distribution of employees who left and stayed differed across their categories.

#### Categorical features EDA summary

Categorical feature analysis identified OverTime, JobRole, BusinessTravel, MaritalStatus, Department, and EducationField as features significantly associated with employee attrition. These findings were supported by the attrition-rate bar charts and Chi-square tests. In contrast, Gender showed no statistically significant association with attrition.

```python
# Model dataframe
# Remove non-predictive variables
X = df_more_1_val.drop(
    columns=[
        # Target, ID, and constant Coulmns
        "Attrition", #Target
        "Attrition_Flag", #Target
        "EmployeeNumber", #ID
        "DailyRate",
        "HourlyRate",
        "MonthlyRate"
    ],
    errors="ignore"
)

y = df_more_1_val["Attrition_Flag"]
```

#### <b> One-Hot Encode Categorical Variables </b>

```python
# One-hot encoding for categorical columns
X_encoded = pd.get_dummies(
    X,
    drop_first=True
)
```

```python
X_encoded.shape
```

#### <b> Train-Test Split </b>

```python
# Split train and test set
from sklearn.model_selection import train_test_split

# Encode categorical variables first
X_encoded = pd.get_dummies(X, drop_first=True)

# Split encoded data
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y #target is imbalanced, Attrition No  = many, Yes = fewer
)
```

### <b> Part 3.2: Baseline Model </b>

#### Train the Logistic Regression Pipeline

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, roc_auc_score

# Build Logistic Regression pipeline
log_pipeline = Pipeline([
    
    # Standardize numeric features
    ("scaler", StandardScaler()),
    
     # Train Logistic Regression model
    ("model", LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ))
])

# Train the pipeline
log_pipeline.fit(X_train, y_train)
```

#### Evaluate model

```python
from sklearn.metrics import classification_report, roc_auc_score

y_pred = log_pipeline.predict(X_test)
y_prob = log_pipeline.predict_proba(X_test)[:, 1]

print(classification_report(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))
```

#### <b> Baseline Model Summary </b>
The baseline logistic regression model achieved an ROC-AUC score of 0.81, indicating a moderate ability to distinguish between employees who stayed and employees who left. For the attrition class, recall was 0.68, meaning the model identified 68% of actual leavers. However, precision was low at 0.39, meaning many employees predicted as likely to leave did not actually leave. The F1-score for the attrition class was 0.49, showing that the model's performance for predicting leavers is still limited.

Although the overall accuracy was 0.69, accuracy is less reliable in this case because the datast is imbalanced, with more employees who stayed  than who left. Therefore, recall, precision, F1-score, and ROC-AUC are more useful for evaluating the model.
 

### <b> Part 3.3: Alternative Models </b>

#### Random Forest Model

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score

rf_model = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    class_weight="balanced"
)

rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)
y_prob_rf = rf_model.predict_proba(X_test)[:, 1]

print(classification_report(y_test, y_pred_rf))
print("ROC-AUC:", roc_auc_score(y_test, y_prob_rf))
```

#### <b> Gradient Boosting Model </b>

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import classification_report, roc_auc_score

gb_model = GradientBoostingClassifier(
    random_state=42
)

gb_model.fit(X_train, y_train)

y_pred_gb = gb_model.predict(X_test)
y_prob_gb = gb_model.predict_proba(X_test)[:,1]

print(classification_report(y_test, y_pred_gb))
print("ROC-AUC:", roc_auc_score(y_test, y_prob_gb))
```

### <b> Part 3.4: Model Comparison and Generalisation Assessment </b>

#### <b> Model Comparison </b>

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

def evaluate_model(model_name, y_test, y_pred, y_prob):
    return {
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision_Attrition_Yes": precision_score(y_test, y_pred),
        "Recall_Attrition_Yes": recall_score(y_test, y_pred),
        "F1_Attrition_Yes": f1_score(y_test, y_pred),
        "ROC_AUC": roc_auc_score(y_test, y_prob)
    }

model_comparison = pd.DataFrame([
    evaluate_model("Baseline Scaled Logistic Regression", y_test, y_pred, y_prob),
    evaluate_model("Random Forest", y_test, y_pred_rf, y_prob_rf),
    evaluate_model("GB model", y_test, y_pred_gb, y_prob_rf)
])

model_comparison
```

Although Gradient Boosting achieved the highest accuracy (0.850) and precision (0.588), its recall was substantially lower than Logistic Regression (0.213 vs 0.574). Since the primary objective was to identify employees at risk of attrition, recall and F1-score were prioritised over overall accuracy. Consistently, Logistic Regression was selected for further threshold optimisation.

#### <b> Overfitting Assessment </b>

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
import pandas as pd

def evaluate_train_test(model_name, model, X_train, X_test, y_train, y_test):
    
    # Train predictions
    y_train_pred = model.predict(X_train)
    y_train_prob = model.predict_proba(X_train)[:, 1]
    
    # Test predictions
    y_test_pred = model.predict(X_test)
    y_test_prob = model.predict_proba(X_test)[:, 1]
    
    return {
        "Model": model_name,
        
        "Train_Accuracy": accuracy_score(y_train, y_train_pred),
        "Test_Accuracy": accuracy_score(y_test, y_test_pred),
        
        "Train_Recall_Yes": recall_score(y_train, y_train_pred),
        "Test_Recall_Yes": recall_score(y_test, y_test_pred),
        
        "Train_F1_Yes": f1_score(y_train, y_train_pred),
        "Test_F1_Yes": f1_score(y_test, y_test_pred),
        
        "Train_ROC_AUC": roc_auc_score(y_train, y_train_prob),
        "Test_ROC_AUC": roc_auc_score(y_test, y_test_prob)
    }
```

```python
overfit_check = pd.DataFrame([
    evaluate_train_test(
        "Scaled Logistic Regression",
        log_pipeline,
        X_train,
        X_test,
        y_train,
        y_test
    ),
    evaluate_train_test(
        "Random Forest",
        rf_model,
        X_train,
        X_test,
        y_train,
        y_test
    ),
    evaluate_train_test(
        "Gradient Boosting",
        gb_model,
        X_train,
        X_test,
        y_train,
        y_test
    )
])

overfit_check
```

The overfitting assessment compared model performance on the training and test sets. Random Forest achieved perfect training performance but substantially lower test performance, particularly for the attrition class, indicating severe overfitting. Gradient Boosting also showed considerable declines in recall, F1-score, and ROC-AUC from training to testing, suggesting considerable overfitting. In contrast, Scaled Logistic Regression exhibited much smaller train–test gaps and maintained the strongest recall on the test set, indicating better generalisation to unseen employee data than the tree-based models.

### <b> Part 3.5: Final Model Optimisation </b>

#### <b> Prediction Threshold Tuning </b>

```python
from sklearn.metrics import precision_score, recall_score, f1_score

threshold_results = []

# Scale + train logistic regression
for threshold in [0.3, 0.4, 0.5, 0.6, 0.7, 0.8]:
    y_pred_threshold = (y_prob >= threshold).astype(int)
    
    threshold_results.append({
        "Threshold": threshold,
        "Precision": precision_score(y_test, y_pred_threshold),
        "Recall": recall_score(y_test, y_pred_threshold),
        "F1_score": f1_score(y_test, y_pred_threshold)
    })
    
threshold_df = pd.DataFrame(threshold_results)
threshold_df
```

```python
# Predict probabilities using the trained logistic regression model
y_prob_06_threshold = log_pipeline.predict_proba(X_test)[:, 1]

selected_threshold = 0.6

# Convert probabilities into binary predictions
y_pred_06_threshold = (
    y_prob_06_threshold >= selected_threshold
).astype(int)

print(classification_report(y_test, y_pred_06_threshold))
print("ROC-AUC:", roc_auc_score(y_test, y_prob_06_threshold))
```

#### <b> Threshold Tuning Summary </b>

The default decision threshold of Logistic Regression (0.5) was adjusted to evaluate the trade-off between precision and recall for predicting employee attrition. Lower thresholds increased recall by identifying more employees who were likely to leave but also produced more false positives, whereas higher thresholds improved precision at the expense of recall.

Among the evaluated thresholds, 0.6 achieved the best overall performance, with the highest F1-score (0.495) while maintaining a balanced precision (43.5%) and recall (57.4%). Therefore, a threshold of 0.6 was selected as the final decision threshold because it provided the most balanced classification performance for this study.
